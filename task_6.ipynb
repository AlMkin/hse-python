```python
import csv

visit_log_path = 'visit_log.csv'
funnel_path = 'funnel.csv'

with open(visit_log_path, 'r', encoding='utf-8') as infile, open(funnel_path, 'w', encoding='utf-8', newline='') as outfile:
    reader = csv.reader(infile)
    writer = csv.writer(outfile)

    header = next(reader)
    writer.writerow(header)

    for row in reader:
        if len(row) > 2 and row[2].strip():
            writer.writerow(row)

print(f"Отфильтрованные данные были записаны в {funnel_path}")
```
