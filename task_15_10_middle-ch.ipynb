```python
def get_middle_characters(word):
    length = len(word)
    middle_index = length // 2
    
    if length % 2 == 0:
        return word[middle_index - 1:middle_index + 1]
    else:
        return word[middle_index]

word1 = 'test'
word2 = 'testing'

print(get_middle_characters(word1))
print(get_middle_characters(word2))
```
