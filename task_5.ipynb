```python
from datetime import datetime

def parse_date(input_date):
    formats = [
        "%A, %B %d, %Y",
        "%A, %d.%m.%y",
        "%A, %d %B %Y"
    ]

    for fmt in formats:
        try:
            return datetime.strptime(input_date, fmt)
        except ValueError:
            continue

    return None

def main():
    print("Введите дату в формате газеты или 'q' для выхода.")
    while True:
        user_input = input("Введите дату: ").strip()
        if user_input.lower() == 'q':
            print("Программа завершена.")
            break

        parsed_date = parse_date(user_input)
        if parsed_date:
            print(f"Объект datetime: {parsed_date}")
        else:
            print("Не удалось распознать дату. Попробуйте снова.")

if __name__ == "__main__":
    main()
```
